```python
import pandas as pd
```

Read in datasets for review and cleaning.

First data set, lego_sets_and_themes.csv from https://www.kaggle.com/datasets/jkraak/lego-sets-and-themes-database

Loading data, dropping missing values showing head, list of column names and rows remaining.  And export the cleaned data back out to csv.

```python
data_themes = pd.read_csv(r'data\lego_sets_and_themes.csv')
                          
cleaned_data_themes = data_themes.dropna()

#removing the -1 at the end of the set_number

cleaned_data_themes['set_number'] = cleaned_data_themes['set_number'].astype(str).str.replace('-1$', '', regex=True)

print(cleaned_data_themes.head())


column_names_themes = cleaned_data_themes.columns.tolist()  # Convert to a list for easier readability
print("Themes column names:")
print(column_names_themes)

cleaned_data_themes.shape[0]



#cleaned_data_themes.to_csv('data\cleaned_lego_sets_and_themes.csv', index=False)
```

Second data set, lego_sets.csv from https://www.kaggle.com/datasets/mterzolo/lego-sets

```python
data_sets = pd.read_csv(r'data\lego_sets.csv')

cleaned_data_sets = data_sets.dropna()
#filter so only reviews from US are kept
filtered_data_sets = cleaned_data_sets[cleaned_data_sets['country']== 'US']

#add a new column to calculate the average piece cost of set based on the list price and piece count columns
filtered_data_sets['average_piece_cost'] = filtered_data_sets['list_price'] / filtered_data_sets['piece_count']

print(filtered_data_sets.head())



column_names_sets = filtered_data_sets.columns.tolist()  # Convert to a list for easier readability
print("Sets column names:")
print(column_names_sets)
filtered_data_sets.shape[0]

#filtered_data_sets.to_csv('data\cleaned_lego_sets.csv', index=False)
```

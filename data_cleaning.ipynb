```python
import pandas as pd
import sqlite3
```

Read in datasets for review and cleaning.

First data set, lego_sets_and_themes.csv from https://www.kaggle.com/datasets/jkraak/lego-sets-and-themes-database

Loading data, dropping missing values showing head, list of column names and rows remaining.  And export the cleaned data back out to csv.

```python
data_themes = pd.read_csv(r'data\lego_sets_and_themes.csv')
                          
cleaned_data_themes = data_themes.dropna()


#removing the -1, -40, -2 at the end of the set_number

cleaned_data_themes['set_number'] = cleaned_data_themes['set_number'].astype(str).str.replace('-1$', '', regex=True)
cleaned_data_themes['set_number'] = cleaned_data_themes['set_number'].astype(str).str.replace('-40$', '', regex=True)
cleaned_data_themes['set_number'] = cleaned_data_themes['set_number'].astype(str).str.replace('-2$', '', regex=True)

#convert set_number to numeric from string coercing errors to NaN
cleaned_data_themes['set_number'] = pd.to_numeric(cleaned_data_themes['set_number'], errors='coerce')

# drop rows with NaN values
cleaned_data_themes.dropna(inplace=True)

#change set_number to int
cleaned_data_themes['set_number'].astype(int)

print(cleaned_data_themes.head())


column_names_themes = cleaned_data_themes.columns.tolist()  # Convert to a list for easier readability
print("Themes column names:")
print(column_names_themes)

cleaned_data_themes.shape[0]


#export filtered data to csv to use in next part of project
cleaned_data_themes.to_csv('data\cleaned_lego_sets_and_themes.csv', index=False)
```

Second data set, lego_sets.csv from https://www.kaggle.com/datasets/mterzolo/lego-sets

```python
data_sets = pd.read_csv(r'data\lego_sets.csv')

cleaned_data_sets = data_sets.dropna()
#filter so only reviews from US are kept
filtered_data_sets = cleaned_data_sets[cleaned_data_sets['country']== 'US']

#add a new column to calculate the piece cost of set based on the list price and piece count columns
filtered_data_sets['piece_cost'] = filtered_data_sets['list_price'] / filtered_data_sets['piece_count']

#rename prod_id to set_number
filtered_data_sets.rename(columns={'prod_id': 'set_number'}, inplace=True)

#change set_number to int
filtered_data_sets['set_number'].astype(int)



print(filtered_data_sets.head())



column_names_sets = filtered_data_sets.columns.tolist()  # Convert to a list for easier readability
print("Sets column names:")
print(column_names_sets)
filtered_data_sets.shape[0]

#export filtered data to csv to use in next part of project
filtered_data_sets.to_csv('data\cleaned_lego_sets.csv', index=False)
```

Comparing column names from the current data sets to determine how to use.

```python
print("First data set - themes")
print(column_names_themes)
print("Second data set - sets")
print(column_names_sets)
```

Use data cleaned in data_cleaning.ipynb (cleaned_lego_sets_and_themes.csv & cleaned_lego_sets.csv) to setup sqlite db

First import necessary items

```python


# combine DataFrames based on dataframes that were also exported to .csv.(cleaned_data_themes & filtered_data_sets)
combined_df = pd.merge(cleaned_data_themes, filtered_data_sets, on='set_number')

# create a SQLite database and import the combined Dataframe
conn = sqlite3.connect('combined_data.db')
combined_df.to_sql('combined_table', conn, if_exists='replace', index=False)

#Close the connection
conn.close()

print ('data frames have been imported and combined into SQLite database')
```
